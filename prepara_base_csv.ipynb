```python
import os
import pandas as pd
from pathlib import Path
import json
```

# Leitura do Excel

```python
taco = pd.read_excel(Path(os.getcwd(), "docs", "TACO_Tabel.xlsx"), sheet_name = "Taco", index_col = 0)
taco
```

```python
taco.info()
```

```python
taco = taco.replace("Tr", 0.0)
taco = taco.replace("NA", 0.0)
taco
```

# Filtrando com base nos alimentos pré selecionados

```python
# Lendo o arquivo JSON
with open(Path(os.getcwd(), "infos", "alimentos_selecionados.json"), 'r') as f:
    data = json.load(f)

# Acessando os dados
alimentos_gramas = data['alimentos_gramas']
alimentos_inteiros = data['alimentos_unidade']
gramas_unidade = data["gramas_por_unidade"]
alimentos_gramas, alimentos_inteiros, gramas_unidade
```

```python
df = taco.loc[alimentos_gramas, :]
df["Tipo"] = "gramas"
df
```

```python
df_temp = taco.loc[alimentos_inteiros, :]
df_temp["Tipo"] = "inteiros"
df_temp
```

```python
base_df = pd.concat([df, df_temp])
base_df
```

```python
base_df = base_df.replace("b", "", regex = True)
base_df
```

```python
base_df = base_df.drop(["Preparo", "Energia (Kcal)", "Umidade", "Energia (KJ)", "Cinzas (g)", "RE (μg)", "RAE (μg)"], axis = 1)
base_df
```

```python
df = base_df.drop(["Nome", "Tipo"], axis = 1).replace(',', '.', regex=True).astype(float).copy()
df = df/100 #dividindo por 100 gramas (base das contas)
for k, id_int in enumerate(alimentos_inteiros):
    df.loc[id_int] = df.loc[id_int]*gramas_unidade[k]
df["Nome"] = base_df["Nome"]
df["Tipo"] = base_df["Tipo"]
df
```

```python
df.to_csv(Path(os.getcwd(), "infos", "base_csv.csv"), index = False)
```
